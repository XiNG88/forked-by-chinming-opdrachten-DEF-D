```python

def vlakkeplaat(A, d, eps_r=3.0, eps0=8.8541878128e-12):
    """Capaciteit van een vlakke-plaatcondensator in farad."""
    return eps0 * eps_r * A / d

# Gegeven
A = 0.01                 # m^2
d_papier = 0.1e-3        # 0.1 mm = 1e-4 m per vel papier
eps_r = 3.0

# X-schaal: max 50 vellen papier
vellen = np.arange(1, 51)           # 1 t/m 50 vellen
d = vellen * d_papier               # dikte in meter

# Capaciteit in pF
C_F = vlakkeplaat(A, d, eps_r=eps_r)
C_pF = C_F * 1e12

# Plot
plt.figure(figsize=(8, 5))
plt.plot(vellen, C_pF, marker="o", markersize=3)
plt.xlabel("Aantal vellen papier")
plt.ylabel("Capaciteit (pF)")
plt.title("Capaciteit vs. aantal vellen papier")
plt.xlim(0, 50)      # x-range 0 t/m 50 vellen
plt.ylim(0, 500)     # y-range 0 t/m 500 pF
plt.xticks(np.arange(0, 51, 5))
plt.yticks(np.arange(0, 501, 50))
plt.grid(True)
plt.tight_layout()
plt.show()

# Optioneel: waarden binnen de plotrange tonen
print("Waarden binnen 0-500 pF:")
for n, c in zip(vellen, C_pF):
    if 0 <= c <= 500:
        print(f"{n:2d} vellen -> {c:8.2f} pF")
```

## Opdracht 7: Plot de kalibratiegrafiek.

```python
# Plot hier je gemeten kalibratiegrafiek en plot in dezelfde figuur ook je verwachte ontwerpgrafiek.

eps0 = 8.8541878128e-12   # F/m
eps_r = 3.0
A = 0.01                  # m^2
d_papier = 0.1e-3         # m per vel


n = np.array([1, 2, 3, 4, 5, 10, 15, 20, 25, 30, 35], dtype=float)
C_pF = np.array([500, 350, 325, 260, 225, 140, 110, 70, 65, 58, 53], dtype=float)

# 2% onzekerheid
C_err_pF = 0.02 * C_pF


C_uF = C_pF * 1e-6
C_err_uF = C_err_pF * 1e-6


def C_model_uF(n, a, b):
    d = n * d_papier  # meter
    d_fit = (d ** b) + a
    C_F = eps0 * eps_r * A / d_fit
    return C_F * 1e6  # naar microfarad


p0 = [1e-4, 1.0]   # [a, b]

bounds_lower = [0.0, 0.1]
bounds_upper = [1.0, 5.0]

popt, pcov = curve_fit(
    C_model_uF,
    n, C_uF,
    p0=p0,
    sigma=C_err_uF,
    absolute_sigma=True,
    bounds=(bounds_lower, bounds_upper),
    maxfev=20000
)

a_fit, b_fit = popt
a_err, b_err = np.sqrt(np.diag(pcov))

print("Fitresultaat voor C = eps0*eps_r*A / ((d**b)+a)")
print(f"a = {a_fit:.6e} ± {a_err:.2e}  (eenheid: m^b)")
print(f"b = {b_fit:.6f} ± {b_err:.4f}")

C_pred_uF = C_model_uF(n, a_fit, b_fit)
res = C_uF - C_pred_uF


# R^2 (niet perfect voor gewogen niet-lineaire fit, maar handig als indicatie)
ss_res = np.sum((C_uF - C_pred_uF)**2)
ss_tot = np.sum((C_uF - np.mean(C_uF))**2)
r2 = 1 - ss_res / ss_tot


vellen2 = np.arange(1, 51, dtype=float)
C_fit_uF = C_model_uF(vellen2, a_fit, b_fit)
C_fit_pF = C_fit_uF * 1e6  # µF -> pF

plt.figure(figsize=(9, 5.5))

plt.errorbar(
    n, C_pF, yerr=C_err_pF,
    fmt='o', capsize=4, elinewidth=1.2,
    label="Metingen (±2%)"
)

plt.plot(
    vellen2, C_fit_pF, '--', linewidth=2,
    label=(fr"Fit: $C=\frac{{\epsilon_0\epsilon_r A}}{{(d^b)+a}}$" "\n"
           fr"$a={a_fit:.2e}$ (m$^b$), $b={b_fit:.3f}$")
)

plt.xlabel("Aantal vellen papier")
plt.ylabel("Capaciteit (pF)")
plt.title("Niet-lineaire fit op $C=\\epsilon_0\\epsilon_r A/((d^b)+a)$")
plt.xlim(0, 50)
plt.ylim(0, 500)
plt.xticks(np.arange(0, 51, 5))
plt.yticks(np.arange(0, 501, 50))
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

## Opdracht 9: Uiteindelijke ontwerp en kalibratiegrafiek.

#### Voeg hier een foto van je uiteindelijke sensor in.

```python
eps0 = 8.8541878128e-12   # F/m
eps_r = 3.0
A = 0.01                  # m^2
d_papier = 0.1e-3         # m per vel

#metingen
n = np.array([1, 2, 3, 4, 5, 10, 15, 20, 25, 30, 35], dtype=float)
C_pF = np.array([500, 350, 325, 260, 225, 140, 110, 70, 65, 58, 53], dtype=float)

# 2% onzekerheid
C_err_pF = 0.02 * C_pF

# pF -> microfarad
# 1 pF = 1e-12 F = 1e-6 µF
C_uF = C_pF * 1e-6
C_err_uF = C_err_pF * 1e-6

def C_model_uF(n, a, b):
    d = n * d_papier  # meter
    d_fit = (d ** b) + a
    C_F = eps0 * eps_r * A / d_fit
    return C_F * 1e6  # naar microfarad

# Theoretisch model (output in microfarad)
# C = eps0 * eps_r * A / d, met d = n * d_papier
def C_theorie_uF(n):
    d = n * d_papier
    C_F = eps0 * eps_r * A / d
    return C_F * 1e6  # naar microfarad

# Fit met curve_fit
# Beginwaarden (belangrijk voor niet-lineaire fit)
# a heeft eenheid m^b (want het wordt opgeteld bij d^b)
p0 = [1e-4, 1.0]   # [a, b]

# Bounds:
# - b positief houden
# - a positief houden voor stabiliteit (anders kan noemer nul/negatief worden)
bounds_lower = [0.0, 0.1]
bounds_upper = [1.0, 5.0]

popt, pcov = curve_fit(
    C_model_uF,
    n, C_uF,
    p0=p0,
    sigma=C_err_uF,
    absolute_sigma=True,
    bounds=(bounds_lower, bounds_upper),
    maxfev=20000
)

a_fit, b_fit = popt
a_err, b_err = np.sqrt(np.diag(pcov))

print("Fitresultaat voor C = eps0*eps_r*A / ((d**b)+a)")
print(f"a = {a_fit:.6e} ± {a_err:.2e}  (eenheid: m^b)")
print(f"b = {b_fit:.6f} ± {b_err:.4f}")

# Fitkwaliteit (op meetpunten)
C_pred_uF = C_model_uF(n, a_fit, b_fit)
res = C_uF - C_pred_uF

# R^2 (niet perfect voor gewogen niet-lineaire fit, maar handig als indicatie)
ss_res = np.sum((C_uF - C_pred_uF) ** 2)
ss_tot = np.sum((C_uF - np.mean(C_uF)) ** 2)
r2 = 1 - ss_res / ss_tot
print(f"R^2 = {r2:.4f}")

# Plot (in pF voor herkenbare schaal)
vellenfinal = np.arange(1, 51, dtype=float)

# Fitlijn
C_fit_uF = C_model_uF(vellenfinal, a_fit, b_fit)
C_fit_pF = C_fit_uF * 1e6  # µF -> pF

# Theoretische lijn
C_theorie_plot_uF = C_theorie_uF(vellenfinal)
C_theorie_plot_pF = C_theorie_plot_uF * 1e6  # µF -> pF

plt.figure(figsize=(9, 5.5))

plt.errorbar(
    n, C_pF, yerr=C_err_pF,
    fmt='o', capsize=4, elinewidth=1.2,
    label="Metingen (±2%)"
)

plt.plot(
    vellenfinal, C_fit_pF, '--', linewidth=2,
    label=(fr"Fit: $C=\frac{{\epsilon_0\epsilon_r A}}{{(d^b)+a}}$" "\n"
           fr"$a={a_fit:.2e}$ (m$^b$), $b={b_fit:.3f}$")
)

# Toegevoegde theoretische plotlijn
plt.plot(
    vellenfinal, C_theorie_plot_pF, '-', linewidth=2,
    label=r"Theorie: $C=\frac{\epsilon_0\epsilon_r A}{d}$ met $d=n\,d_{papier}$"
)

plt.xlabel("Aantal vellen papier")
plt.ylabel("Capaciteit (pF)")
plt.title("capaciteit tegen aantal vellen met metingen/fit en theoretische verwachting")
plt.xlim(0, 50)
plt.ylim(0, 500)
plt.xticks(np.arange(0, 51, 5))
plt.yticks(np.arange(0, 501, 50))
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```
