##### Antwoorden 4b

1.
We nemen twee parallelle platen loodrecht op de $z$-as, met vaste potentialen:

bovenste plaat op $z = +d/2$: $V_1$
onderste plaat op $z = -d/2$: $V_2$

we nemen aan dat het ideale platen zijn die oneindig groot zijn

Door symmetrie hangt het potentiaal alleen af van $z$: $V = V(z)$ en het elektrische veld heeft alleen een $z$-component: $\vec E = E_z(z)\,\hat z$.

In het algemeen geldt:
$$
\vec E = -\nabla V
$$
Omdat $V$ alleen van $z$ afhangt, wordt dit:
$$
E_z = -\frac{dV}{dz}
$$

Tussen de platen zijn er geen vrije ladingen, dus geldt Laplace:
$$
\nabla^2 V = 0
$$
In 1D:
$$
\frac{d^2V}{dz^2} = 0
$$
Integreer twee keer:
$$
\frac{dV}{dz} = A,\qquad V(z)=Az+B
$$

Randvoorwaarden:
$$
V\!\left(+\frac{d}{2}\right)=V_1,\qquad V\!\left(-\frac{d}{2}\right)=V_2
$$

Invullen:
$$
V_1 = A\left(\frac{d}{2}\right)+B,\qquad V_2 = A\left(-\frac{d}{2}\right)+B
$$

Aftrekken geeft:
$$

1. Vind de formule voor de capaciteit van een vlakke plaat condensator en een cilinder condensator, onder de aanname dat de afstand tussen de platen veel kleiner is dan het oppervlakte van de condensator (vlakke plaat aanname).

````python
import numpy as np

#in 5a hebben we een functie gemaakt die de capaciteit berekend van een condensator
# op basis van de geometrie en het materiaal

# 2) Vlakke plaat condensator (parallel plates)
# Formule: C = ε0 * εr * A / d
# A = oppervlakte overlap (m^2), d = plaatafstand (m)

# Capaciteit ideale vlakke plaat condensator
# Input:
# A = overlap-oppervlakte (m^2)
# d = afstand tussen platen (m)
#eps_r = relatieve permittiviteit (-), standaard 1 (lucht/vacuüm)
# eps0  = permittiviteit vacuüm (F/m)
# Output:
# C = capaciteit (Farad)
def vlakkeplaat(A, d, eps_r=1.0, eps0=8.8541878128e-12):
    return eps0 * eps_r * A / d


# 4) Cilindercondensator (coaxiaal)
# Exacte formule: C = 2π ε0 εr L / ln(b/a)
# a = binnenstraal, b = buitenstraal, L = lengte
# 3) Kleine spleet-approx (b-a << a): ln(b/a) ≈ (b-a)/a  -> C ≈ 2π ε0 εr L * a / (b-a)
# Capaciteit coaxiale cilindercondensator
# Input:
# a = binnenstraal (m)
# b = buitenstraal (m), moet b > a zijn
# L = lengte (m)
# eps_r = relatieve permittiviteit (-), standaard 1
#eps0 = permittiviteit vacuüm (F/m)
# small_gap= True > kleine-spleet benadering (b-a << a), vaak niet echt nodig
#  False > exacte formule
# Output:
#  C  = capaciteit (Farad)
def cilinder(a, b, L, eps_r=1.0, eps0=8.8541878128e-12, small_gap=False):
    if b <= a:
      raise ValueError("Voor een cilindercondensator moet b > a zijn.")
    if small_gap:
        gap = b - a
        return 2 * np.pi * eps0 * eps_r * L * a / gap
    else:
        return 2 * np.pi * eps0 * eps_r * L / np.log(b / a)
````

### Vraag 5a: Formules voor de capaciteit (en waarom)

Vlakke plaatcondensator  
Bij twee grote, parallelle platen met afstand $d$ (waarbij $d$ veel kleiner is dan de afmetingen van de platen) is het elektrische veld tussen de platen bijna uniform. In dat ideale geval geldt:
$$
C = \varepsilon_0 \varepsilon_r \frac{A}{d}
$$
Hierin is $A$ het overlap-oppervlak en $\varepsilon_0\varepsilon_r$ de permittiviteit van het medium tussen de platen. Deze formule kiezen we omdat de capaciteit groter wordt bij meer overlap ($A$) en kleiner wordt bij grotere afstand ($d$).

Cilindercondensator (coaxiaal)  
Voor een binnenstraal $a$, buitenstraal $b$ en lengte $L$ is de exacte capaciteit:
$$
C = \frac{2\pi \varepsilon_0 \varepsilon_r L}{\ln(b/a)}
$$
Deze formule kiezen we omdat het elektrische veld radiaal is en met $1/r$ verandert, waardoor de log-term ontstaat.

Als de spleet klein is ($b-a \ll a$), dan geldt de benadering $\ln(b/a)\approx (b-a)/a$ en wordt:
$$
C \approx 2\pi \varepsilon_0 \varepsilon_r L \frac{a}{b-a}
$$
Deze vorm laat duidelijk zien dat de capaciteit groter wordt als de afstand $(b-a)$ kleiner wordt.

Het procentuele verschil nemen we als:

de disk-factor is steeds iets groter dan de square-factor.

#### Vraag 6b2: Capaciteit van een condensator met het edge effect meegenomen

Vanaf hier nemen we aan dat factor van de circulaire platen ongeveer gelijk is aan de factor voor vierkante platen. De factor die de capaciteit van de vlakke plaat approximatie linkt aan de werkelijke capaciteit noemen we vanaf nu de idealiteitsfactor.

Vanaf hier gaan we verder met jullie sensor van de ontwerpopdracht "Maak een sensor met condensatoren":

3. Reken je theoretische kalibratie grafiek om met de idealiteitsfactor.
4. Vergelijk je nieuwe theoretische kalibratiegrafiek van je praktische kalibratiegrafiek van de dag zelf.

````python
# Schrijf hier jouw code die de vlakke plaat capaciteit omschrijft naar de werkelijke capaciteit

# def C_niet_ideaal(#relevante parameters):
    # Jouw berekeningen

# 1 Rekent de "sanity check 3"-condensator (uit DEFONT) door:
# vlakke-plaat capaciteit C0
# idealiteitsfactor η (op basis van disk-benadering)
# werkelijke (gecorrigeerde) capaciteit C = η*C0
# vergelijkt met de gemeten capaciteit van de practicumdag
# 2 Past dezelfde correctie toe op jullie sensor-kalibratie:
# maakt een gecorrigeerde theoretische kalibratiecurve
# plot die samen met jullie gemeten kalibratiepunten

import numpy as np
import matplotlib.pyplot as plt

# Vacuum permittivity (constante)
eps0 = 8.8541878128e-12  # [F/m]

# Idealiteitsfactor (disk-fit)
def idealiteitsfactor_disk_benadering(A, d):
    """
    Berekent idealiteitsfactor η(b) via een disk-benadering.
    We nemen aan dat de factor voor vierkante platen ~ gelijk is aan die van een disk.
    We zetten de plaatoppervlakte A om naar een equivalente schijf met dezelfde A.
    Voor een schijf geldt: R = sqrt(A/pi)
    Aspect ratio: b = d/(2R)
    Empirische lineaire fit (uit paper): η ≈ 1 + 2.5465*b

    Parameters:
    A: oppervlak van één plaat [m^2]
    d: afstand tussen de platen [m]

    Returns:
    - η: dimensieloos (>=1), vermenigvuldigingsfactor voor C0
    """
    R = np.sqrt(A / np.pi)     # equivalente schijfstraal [m]
    b = d / (2 * R)            # aspect ratio (disk) [-]
    eta = 1.0 + 2.5465 * b     # lineaire fit [-]
    return eta

def C_vlakkeplaat(A, d, eps_r=1.0):
    """
    Vlakke-plaat benadering:
    C0 = eps0*eps_r*A/d

    Parameters:
    A: oppervlak [m^2]
    d: afstand [m]
    eps_r: relatieve permittiviteit [-] (lucht ~1)

    Returns:
    C0 [F]
    """
    return eps0 * eps_r * A / d

def C_niet_ideaal(A, d, eps_r=1.0):
    """
    Werkelijke capaciteit met edge-effect (benadering):
    C ≈ η * C0

    Returns:
    - C [F]
    """
    C0 = C_vlakkeplaat(A, d, eps_r)              # vlakke-plaat capaciteit [F]
    eta = idealiteitsfactor_disk_benadering(A, d) # idealiteitsfactor [-]
    return eta * C0

# 1 Sanity check
A_sc = 0.1            # [m^2] oppervlak (sanity check 3)
d_sc = 0.0025         # [m] plaatafstand (sanity check 3)
eps_r_sc = 1.0        # [-] lucht (pas aan als er een dielectricum gebruikt is)
C_meet_sc = 0.9947e-10 # [F] gemeten capaciteit (sanity check 3)

# Berekeningen voor sanity check 3
C0_sc = C_vlakkeplaat(A_sc, d_sc, eps_r_sc)      # vlakke-plaat [F]
eta_sc = idealiteitsfactor_disk_benadering(A_sc, d_sc) # idealiteitsfactor [-]
C_sc = C_niet_ideaal(A_sc, d_sc, eps_r_sc)       # gecorrigeerd [F]

# Netjes printen (in pF)
print("=== Sanity check 3 ===")
print(f"A = {A_sc:.3g} m^2, d = {d_sc:.4g} m, eps_r = {eps_r_sc}")
print(f"C_vlakkeplaat (C0)     = {C0_sc*1e12:.2f} pF")
print(f"Idealiteitsfactor (η)  = {eta_sc:.5f}")
print(f"C_werkelijk (η*C0)     = {C_sc*1e12:.2f} pF")
print(f"C_gemeten              = {C_meet_sc*1e12:.2f} pF")

# Relatief verschil: hoe ver zit de meting van de theorie (gecorrigeerd)
rel_diff_pct = (C_sc - C_meet_sc) / C_sc * 100.0
print(f"Relatief verschil (theorie vs meting) = {rel_diff_pct:.1f} %")

# 2 kalibratiecurve corrigeren met η

eps_r = 3.0  # [-] relatieve permittiviteit van papier/stack 
A = 0.01    # [m^2] plaatoppervlak van sensor 
d_papier = 0.1e-3  # [m] dikte van 1 vel papier 

# Gemeten kalibratiepunten : n = aantal vellen, C_pF = capaciteit in pF
n_meet = np.array([1, 2, 3, 4, 5, 10, 15, 20, 25, 30, 35], dtype=float)
C_meet_pF = np.array([500, 350, 325, 260, 225, 140, 110, 70, 65, 58, 53], dtype=float)

# Onzekerheid van metingen: 2% relatieve fout
C_meet_err_pF = 0.02 * C_meet_pF

def d_van_n(n):
    """
    Zet aantal vellen om naar plaatafstand d(n).
    """
    return n * d_papier

def C_theorie_vlakkeplaat(n):
    """
    Theoretische capaciteit (vlakke-plaat) als functie van n.
    """
    d = d_van_n(n)
    return C_vlakkeplaat(A, d, eps_r)

def C_theorie_gecorrigeerd(n):
    """
    Theoretische capaciteit met idealiteitsfactor als functie van n:
    C_corr(n) = η(b(n)) * C0(n)
    """
    d = d_van_n(n)
    return C_niet_ideaal(A, d, eps_r)

# Maak een 'gladde' curve voor de plot
n_fijn = np.arange(1, 51, dtype=float)

# Zet naar pF voor leesbare grafiek
C_th_pF = C_theorie_vlakkeplaat(n_fijn) * 1e12
C_th_corr_pF = C_theorie_gecorrigeerd(n_fijn) * 1e12

# Plot: metingen + oude theorie + gecorrigeerde theorie
plt.figure(figsize=(9, 5.5))
plt.errorbar(n_meet, C_meet_pF, yerr=C_meet_err_pF, fmt='o', capsize=4, label="Metingen (±2%)")
plt.plot(n_fijn, C_th_pF, linewidth=2, label="Theorie vlakke plaat")
plt.plot(n_fijn, C_th_corr_pF, '--', linewidth=2, label="Theorie + idealiteitsfactor")
plt.xlabel("Aantal vellen papier (n)")
plt.ylabel("Capaciteit (pF)")
plt.title("Kalibratie: vlakke-plaat vs. idealiteitsfactor-correctie")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
````

##### Antwoorden 6b2


Vlakke-plaat benadering:
$$C_0=\epsilon_0\epsilon_r\frac{A}{d}$$

Idealiteitsfactor (disk-benadering):
$$R=\sqrt{\frac{A}{\pi}},\quad b=\frac{d}{2R},\quad \eta(b)\approx 1+2.5465\,b$$

Werkelijke capaciteit (met edge effect):
$$C_{\text{werkelijk}}\approx \eta\,C_0$$

Vergelijk met meting:
$$\%\Delta=\frac{C_{\text{werkelijk}}-C_{\text{gemeten}}}{C_{\text{werkelijk}}}\cdot 100\%$$

Sensor (kalibratie met papier):
$$d(n)=n\,d_{\text{papier}},\quad C_{\text{corr}}(n)=\eta(b(n))\,C_0(n)$$


Voor de sanity check 3 berekenen we eerst de vlakke-plaat capaciteit met de geometrie van de condensator. Daarna corrigeren we deze waarde met de idealiteitsfactor (edge effect), waarbij we aannemen dat de factor voor een vierkante plaat ongeveer gelijk is aan die van een circulaire schijf. De zo berekende “werkelijke theoretische capaciteit” vergelijken we direct met de capaciteit die op de dag zelf is gemeten door het (relatieve) verschil te bepalen.

Voor de sensor doen we hetzelfde: we nemen de oude theoretische kalibratiecurve (vlakke-plaat benadering) en vermenigvuldigen elk punt met dezelfde idealiteitsfactor die hoort bij de bijbehorende plaatafstand. Hierdoor krijgen we een nieuwe theoretische kalibratiecurve die het edge effect meeneemt. Die nieuwe curve vergelijken we vervolgens met de praktische kalibratiecurve van de metingen; het verschil is meestal klein omdat de idealiteitsfactor in onze meetrange dicht bij 1 ligt. Waarschijnlijk veroorzaken systematische fouten het verschil, zoals lucht tussen vellen papier en geschatte dichtheid.
